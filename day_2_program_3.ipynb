### PROG 3.1: Compute Circumference and Area of Circle

##### Description - This program calculates the circumference and area of a circle by taking the radius as input, converting it from inches to centimeters, and performing the required mathematical calculations.

```python
# Step 1: Define Python Constant for PI
PI = 22/7

# Step 2: Take input for radius of the circle in inches
radius_inch = float(input("Enter the radius of the circle in inches: "))

# Step 3: Convert from inch to cm by using 1 in = 2.54 cm
radius_cm = radius_inch * 2.54

# Step 4: Compute circumference and area in cm and sqcm. Here we use
# Operator Precedence PEMDAS to (radius_cm ** 2) first
circumference_cm = 2 * PI * radius_cm
area_square_cm = PI * (radius_cm ** 2)

# Step 5: Print the results with new line separator
print(f"\nCircumference of the Circle is {circumference_cm} cm and", f"Area of the Circle is {area_square_cm} sqcm", sep="\n")
```

### Prog 3.2: Compute Circumference and Area of Circle with Format Specifier

##### Description - This program performs the same calculations as PROG 3.1 but formats the output using Python's floating-point format specifier (`:.2f`) to display the results with two decimal places.

```python
# Python Format Specifier for floating-point number

# Step 1: Define Python Constant for PI
PI = 22/7

# Step 2: Take input for radius of the circle in inches
radius_inch = float(input("Enter the radius of the circle in inches: "))

# Step 3: Convert from inch to cm by using 1 in = 2.54 cm
radius_cm = radius_inch * 2.54

# Step 4: Compute circumference and area in centimeters and square centimeters
circumference_cm = 2 * PI * radius_cm
area_square_cm = PI * (radius_cm ** 2)

# Step 5: Print the results
print(f"\nCircumference of the Circle is {circumference_cm:.2f} cm and", f"Area of the Circle is {area_square_cm:.2f} sqcm")
```
