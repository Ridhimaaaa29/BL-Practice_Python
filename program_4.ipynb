Function parameter type annotations

```python
# defining proper function parameter data type
def calculate_sum(x: int, y: int):
    """This function takes two integers and returns their sum."""
    return x + y

total_sum = 10
bonus_points = 20
print(f"Data Type of total_sum is {type(total_sum)} and \n" + f"Data Type of bonus_points is {type(bonus_points)}")
final_score = calculate_sum(total_sum, bonus_points)
print(f"Final score is : {final_score}")
```
