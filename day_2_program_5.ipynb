### Prog 5: Time Conversion

##### Description - This program converts a given time duration in seconds into hours, minutes, and seconds using floor division (`//`) and modulo (`%`) operators.

```python
# Step 1: Take input for time duration in total seconds
total_seconds = int(input("Enter the time duration in seconds: "))

# Step 2: Convert seconds to hours, minutes, and seconds using floor division and modulo operator
hours = total_seconds // 3600
minutes = (total_seconds % 3600) // 60
seconds = total_seconds % 60

# Print the converted time duration
print(f"\nTime duration of {total_seconds} seconds in HH:MM:SS format is", f"{hours:02d}:{minutes:02d}:{seconds:02d}")
```
