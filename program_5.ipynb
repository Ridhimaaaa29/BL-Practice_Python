Hard Coding Values

```python
# Specifying a Global Constant for PI
PI =3.14

def calculate_area(radius):
    return PI * radius ** 2

radius = 2
print(f"Area of circle with radius {radius} is : {calculate_area(radius)}")
```

Remove Hard Coding 

```python
# Using user input for total_sum and bonus_points
def calculate_sum(x: int, y: int):
    """This function takes two integers and returns their sum."""
    return x + y

total_sum = input ("Enter the total sum: ")
bonus_points = input ("Enter the bonus points: ")
print(f"Data Type of total_sum is {type(total_sum)} and \n" + f"Data Type of bonus_points is {type(bonus_points)}")
final_score = calculate_sum(total_sum, bonus_points)
print(f"Final score is : {final_score}")
```
