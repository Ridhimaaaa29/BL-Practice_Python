##### Prog 4 - Formatted String Literal

##### Problem Statement:
Take your friend's name and place of stay as input. Display the information in the following format using Python's formatted string literal (f-string):

**"Name of my friend is <name> and his/her place of stay is <place>."**

##### Input:
- Friend's name
- Place of stay

##### Output:
- A formatted sentence displaying the friend's name and place of stay.

##### Description
This program demonstrates the use of Python's formatted string literal (f-string). It accepts the friend's name and place of stay as input and inserts these values directly into a formatted sentence for clear and readable output.

```python
# Step 1: Take friend's name as input
friend_name = input("Enter your friend's name: ")

# Step 2: Take place of stay as input
place_of_stay = input("Enter your friend's place of stay: ")

# Step 3: Display the entered inputs
print(f"\nFriend's Name: {friend_name}")
print(f"Place of Stay: {place_of_stay}")

# Step 4: Display the formatted sentence using an f-string
print(f"\nName of my friend is {friend_name} and his/her place of stay is {place_of_stay}.")
```

##### Retail Invoicing App
##### Problem Statement:
A retail shopkeeper wants to generate a bill for a customer. The shop sells two types of items:
- Household Goods (GST = 5%)
- Processed Food (GST = 12%)
The program should accept the buyer's name, price per unit, and quantity of each item, calculate the GST, generate the final bill, and display the total amount using the Indian Rupee symbol (₹).
##### Input:
- Buyer Name
- Household Good Price
- Household Good Quantity
- Processed Food Price
- Processed Food Quantity
##### Output:
- Item-wise bill with GST
- Total Bill Amount
- Rounded Total Amount
##### Description
This program generates a retail invoice by calculating the cost of household goods and processed food purchased by a customer. It computes GST separately for each item, prepares a formatted invoice, and displays both the total amount and the rounded bill amount using the Indian Rupee symbol.

```python
# Step 1: Set GST rates
HOUSEHOLD_GST = 0.05
PROCESSED_FOOD_GST = 0.12

# Step 2: Take input from the user
buyer_name = input("Enter Buyer Name: ")

household_price = float(input("Enter Household Good Price: "))
household_quantity = float(input("Enter Household Good Quantity: "))

processed_food_price = float(input("Enter Processed Food Price: "))
processed_food_quantity = float(input("Enter Processed Food Quantity: "))

# Step 3: Display the entered inputs
print(f"\nBuyer Name: {buyer_name}")
print(f"Household Good Price: ₹{household_price}")
print(f"Household Good Quantity: {household_quantity}")
print(f"Processed Food Price: ₹{processed_food_price}")
print(f"Processed Food Quantity: {processed_food_quantity}")

# Step 4: Calculate prices
household_total = household_price * household_quantity
processed_food_total = processed_food_price * processed_food_quantity

# Step 5: Calculate GST
household_gst = household_total * HOUSEHOLD_GST
processed_food_gst = processed_food_total * PROCESSED_FOOD_GST

# Step 6: Calculate total including GST
household_bill = household_total + household_gst
processed_food_bill = processed_food_total + processed_food_gst

grand_total = household_bill + processed_food_bill
rounded_total = round(grand_total)

# Step 7: Print the invoice
print(f"\nBuyer Name: {buyer_name}")

print("-" * 94)
print(f"| {'Item Code':<18} | {'Price/Unit':<10} | {'#Unit':<6} | {'Price':<10} | {'GST':<10} | {'Total w/ GST':<12} |")
print("-" * 94)

print(f"| {'Household Good':<18} | ₹ {household_price:<8.0f} | {household_quantity:<6} | ₹ {household_total:<8.1f} | ₹ {household_gst:<8.1f} | ₹ {household_bill:<10.1f} |")

print(f"| {'Processed Food':<18} | ₹ {processed_food_price:<8.0f} | {processed_food_quantity:<6} | ₹ {processed_food_total:<8.1f} | ₹ {processed_food_gst:<8.1f} | ₹ {processed_food_bill:<10.1f} |")

print("-" * 94)

print(f"\n{'Total':<65} ₹ {grand_total:.2f}")
print(f"{'Total Round':<65} ₹ {rounded_total:.2f}")

print("-" * 94)
```
