### PROG 7: School Report App

##### Description - This program generates report cards for three students by taking marks in Physics, Chemistry, and Mathematics. It calculates subject-wise percentages, total marks, overall percentage, and displays a formatted report card for each student. It also generates a class summary showing the average marks and percentage for each subject along with the overall class percentage.

```python
# Step 1: Specify Constants
SCHOOL_NAME = "New School Of Learning"
CLASS_NAME = "Class XI"
TOTAL_MARKS_PER_SUBJECT = 50
NUMBER_OF_STUDENTS = 3

# Variables for class summary
total_physics = 0
total_chemistry = 0
total_mathematics = 0
total_marks_all_students = 0

# Process each student
for student_number in range(1, NUMBER_OF_STUDENTS + 1):

    # Step 2: Get student details
    student_name = input(f"Enter name of student {student_number}: ")

    physics_marks = int(input(f"Enter Physics marks out of 50 for student {student_number}: "))

    chemistry_marks = int(input(f"Enter Chemistry marks out of 50 for student {student_number}: "))

    mathematics_marks = int(input(f"Enter Mathematics marks out of 50 for student {student_number}: "))

    # Step 3: Calculate totals
    total_marks = physics_marks + chemistry_marks + mathematics_marks

    physics_percentage = (physics_marks / TOTAL_MARKS_PER_SUBJECT) * 100
    chemistry_percentage = (chemistry_marks / TOTAL_MARKS_PER_SUBJECT) * 100
    mathematics_percentage = (mathematics_marks / TOTAL_MARKS_PER_SUBJECT) * 100

    overall_percentage = (total_marks / 150) * 100

    # Add for class summary
    total_physics += physics_marks
    total_chemistry += chemistry_marks
    total_mathematics += mathematics_marks
    total_marks_all_students += total_marks

    # Step 4: Print report card
    print()
    print(f"{SCHOOL_NAME} - {CLASS_NAME} - {student_name}")
    print("-" * 72)

    print(f"| {'Subject':^15} | {'Total Marks':^12} | {'Marks Obtained':^15} | {'Percentage':^12} |")

    print("-" * 72)

    print(f"| {'Physics':^15} | {TOTAL_MARKS_PER_SUBJECT:^12} | {physics_marks:^15} | {physics_percentage:^12.2f} |")

    print(f"| {'Chemistry':^15} | {TOTAL_MARKS_PER_SUBJECT:^12} | {chemistry_marks:^15} | {chemistry_percentage:^12.2f} |")

    print(f"| {'Mathematics':^15} | {TOTAL_MARKS_PER_SUBJECT:^12} | {mathematics_marks:^15} | {mathematics_percentage:^12.2f} |")

    print("-" * 72)

    print(f"| {'Total':^15} | {150:^12} | {total_marks:^15} | {overall_percentage:^12.2f} |")

    print("-" * 72)
    print()

# Step 5: Class Summary
physics_average = total_physics / NUMBER_OF_STUDENTS
chemistry_average = total_chemistry / NUMBER_OF_STUDENTS
mathematics_average = total_mathematics / NUMBER_OF_STUDENTS

physics_average_percentage = (physics_average / TOTAL_MARKS_PER_SUBJECT) * 100
chemistry_average_percentage = (chemistry_average / TOTAL_MARKS_PER_SUBJECT) * 100
mathematics_average_percentage = (mathematics_average / TOTAL_MARKS_PER_SUBJECT) * 100

overall_percentage = (total_marks_all_students / (NUMBER_OF_STUDENTS * 150)) * 100

print("Class Average And Percentage for Each Subject:\n")

print(f"Physics Average is {physics_average:.2f} and Percentage is {physics_average_percentage:.2f}%")

print(f"Chemistry Average is {chemistry_average:.2f} and Percentage is {chemistry_average_percentage:.2f}%")

print(f"Mathematics Average is {mathematics_average:.2f} and Percentage is {mathematics_average_percentage:.2f}%")

print(f"\nOverall Percentage: {overall_percentage:.2f}%")
```
